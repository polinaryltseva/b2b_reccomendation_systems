## Из спарка

```python
df = pl.read_excel(source='spark_data/СПАРК-Список_data1.xlsx', 
                   read_options={'header_row':3}, 
                   schema_overrides={'Номер налогоплательщика':pl.Int64})
df.head(2)
```

```python
for i in range(2, 9):
    new = pl.read_excel(source=f'spark_data/СПАРК-Список_data{i}.xlsx', 
                        read_options={'header_row':3}, 
                        schema_overrides={'Номер налогоплательщика':pl.Int64})
    df = pl.concat([df, new])

df.shape
```

```python
nine = pl.read_excel(source=f'spark_data/СПАРК-Список_data9.xlsx', 
                        read_options={'header_row':3}, 
                        schema_overrides={'Номер налогоплательщика':pl.Int64})
df = pl.concat([df, new], how='diagonal')
df.head(1)
```

```python
df = df.select(['Номер налогоплательщика', 'Выручка', 'Сводный риск'])
df = df.with_columns(pl.when(pl.col('Сводный риск').str.contains('Недействующий')).then(pl.lit('Недействующий')).otherwise(pl.col('Сводный риск')).alias('Сводный риск'))
df.head(15)
```

```python
df = df.filter(pl.col('Номер налогоплательщика').is_not_null())
print(df.shape)
df.select(pl.all().null_count())
```

```python
df = df.rename({'Номер налогоплательщика':'ИНН поставщика'})
```

```python
df = df.unique()
df.shape
```

## Соединяем

```python
df2015.shape
```

```python
df2020_rev = df2020.join(other=df, on='ИНН поставщика', how='left')
df2015_rev = df2015.join(other=df, on='ИНН поставщика', how='left')
```

```python
df2020_rev.null_count()
```

```python
df2015_rev.null_count()
```

Сохраним данные без заполнения null

```python
df2015_rev.write_excel( dtype_formats={pl.Date: "yyyy-mm-dd"})
```

Теперь заполним null

```python
df2020_rev = df2020_rev.with_columns(pl.col('Выручка').fill_null(0), pl.col('Сводный риск').fill_null('Риск не определен'))
df2015_rev = df2015_rev.with_columns(pl.col('Выручка').fill_null(0), pl.col('Сводный риск').fill_null('Риск не определен'))
```
